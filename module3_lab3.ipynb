```python
import pandas as pd

bridges = pd.read_csv("bridges.csv")

print("Bridges Dataset:")
print(bridges.head())

print("\nDataset Shape:")
print(bridges.shape)
```

```python
print("Bridge Length Summary Statistics:")
print(bridges["length"].describe())

print("\nMissing Length Values:")
print(bridges["length"].isna().sum())
```

```python
# Assume that bridge lengths between 800 and 3,000 are reasonable for the Pittsburgh bridges dataset.
# I will chose this range because most of the bridge lengths are much smaller than 3,000.
# The minimum recorded length is 804.
# I will investigate the flagged values before deciding whether they should be changed and or removed.

```

```python
# Bridge length range
minimum_length = 800
maximum_length = 3000

# Find bridges outside the range
outliers = bridges[
    (bridges["length"] < minimum_length) |
    (bridges["length"] > maximum_length)
]

print("Possible Bridge Length Outliers:")
print(outliers[["Id", "length", "purpose", "material", "span"]].to_string(index=False))

print("\nTotal Outliers Found:", len(outliers))
```

```python
missing_lengths = bridges[bridges["length"].isna()]

print("Bridges With Missing Lengths:")
print(missing_lengths[["Id", "length"]].to_string(index=False))

print("\nTotal Missing Lengths:", len(missing_lengths))
```

```python
# Create copy of original dataset
cleaned_bridges = bridges.copy()

# Keep flagged lengths unchanged
print("Original Number of Bridges:", len(bridges))
print("Number of Bridges After Review:", len(cleaned_bridges))

print("\nFlagged Values Retained:")
print(cleaned_bridges.loc[
    cleaned_bridges["Id"].isin(["E34", "E46", "E91"]),
    ["Id", "length"]
].to_string(index=False))
```
