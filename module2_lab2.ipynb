```python
def multiply_for(a, b):
    result = 0

    for i in range(b):
        result += a

    return result
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result += a
        count += 1

    return result
```

```python
print("For loop result:", multiply_for(7, 6))
print("While loop result:", multiply_while(7, 6))
print("Expected result:", 42)
```

```python
print(multiply_for(5, 4))
print(multiply_while(5, 4))

print(multiply_for(3, 8))
print(multiply_while(3, 8))

print(multiply_for(0, 6))
print(multiply_while(0, 6))
```

```python
assert multiply_for(7, 6) == 42
assert multiply_while(7, 6) == 42

print("Both functions passed!")
```
