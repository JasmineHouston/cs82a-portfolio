```python

import sqlite3
import pandas as pd

print("SQLite and Pandas imported successfully!")
```

```python

con = sqlite3.connect("customers.db")

print("Database connected successfully!")
```

```python

con.execute("""
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    city TEXT,
    age INTEGER,
    total_spent REAL
)
""")

con.commit()

print("Customers table created successfully!")
```

```python

customers_data = [
    (1, "Ashley Brown", "Los Angeles", 25, 250.50),
    (2, "Michael Smith", "Chicago", 32, 450.75),
    (3, "Jessica Davis", "Houston", 28, 320.00),
    (4, "David Wilson", "New York", 40, 600.25),
    (5, "Sarah Johnson", "Miami", 35, 150.00),
    (6, "James Taylor", "Atlanta", 29, 275.80),
    (7, "Emily Anderson", "Seattle", 31, 510.40),
    (8, "Daniel Martinez", "Dallas", 27, 390.60)
]

con.executemany("""
INSERT OR REPLACE INTO customers
(id, name, city, age, total_spent)
VALUES (?, ?, ?, ?, ?)
""", customers_data)

con.commit()

print("8 fictional customers inserted successfully!")
```

```python

query = "SELECT * FROM customers"

customers = pd.read_sql_query(query, con)

print("Customers Database:")
print(customers.to_string(index=False))
```

```python

query = """
SELECT COUNT(*) AS total_customers
FROM customers
"""

result = pd.read_sql_query(query, con)

print("Total Customers:")
print(result)
```

```python

query = """
SELECT name, city, total_spent
FROM customers
WHERE total_spent > 300
ORDER BY total_spent DESC
"""

result = pd.read_sql_query(query, con)

print("Customers Who Spent More Than $300:")
print(result.to_string(index=False))
```

```python

con.close()

print("Database connection closed successfully!")
```
