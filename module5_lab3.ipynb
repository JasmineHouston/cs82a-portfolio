```python
clean_data = data[
    data["age"].between(0, 120) &
    (data["income"] >= 0) &
    (data["num_vehicles"] >= 0) &
    (data["num_vehicles"] % 1 == 0)
].copy()

print("Original Rows:", len(data))
print("Cleaned Rows:", len(clean_data))
print("Rows Removed:", len(data) - len(clean_data))

print("\nCleaned Customer Data:")
print(clean_data.head())
```

```python
correlation_after = clean_data.corr()

print("Correlation Matrix After Cleaning:")
print(correlation_after.round(3))
```

```python
comparison = pd.DataFrame({
    "Before Cleaning": [
        correlation_before.loc["age", "income"],
        correlation_before.loc["age", "num_vehicles"],
        correlation_before.loc["income", "num_vehicles"]
    ],
    "After Cleaning": [
        correlation_after.loc["age", "income"],
        correlation_after.loc["age", "num_vehicles"],
        correlation_after.loc["income", "num_vehicles"]
    ]
}, index=[
    "Age vs Income",
    "Age vs Vehicles",
    "Income vs Vehicles"
])

comparison["Difference"] = (
    comparison["After Cleaning"] -
    comparison["Before Cleaning"]
)

print("Correlation Comparison:")
print(comparison.round(3))
```

```python
comparison[
    ["Before Cleaning", "After Cleaning"]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Correlation Before and After Data Cleaning")
plt.xlabel("Variable Pairs")
plt.ylabel("Pearson Correlation")
plt.xticks(rotation=0)
plt.ylim(-1, 1)
plt.axhline(0, color="black", linewidth=0.8)
plt.legend()
plt.tight_layout()
plt.show()
```
