```python
# Temperature Categories and Cutoffs

# The temperatures are measured in degrees Celsius (°C).

# Cold: Temperatures below 15°C.
# Warm: Temperatures from 15°C to 25°C, including  both 15°C and 25°C.
# Hot: Temperatures above 25°C.
```

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

print("Weekly Temperatures:")
print(temperatures)
```

```python
for temp in temperatures:
    if ...:
        category = "Cold"
    elif ...:
        category = "Warm"
    else:
        category = "Hot"

    print(f"{temp}°C: {category}")
```
