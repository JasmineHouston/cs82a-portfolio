```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")

print("Dataset Shape:", df.shape)

display(df.head(10))

print("Column Data Types:")
print(df.dtypes)
```

```python
### Data Observations
```

```python
print("Missing Values Before Cleaning:")

print(df.isna().sum())
```

```python
median_price = df["price"].median()

print("Median Price:", median_price)

df["price"] = df["price"].fillna(median_price)

print("Missing Values After Filling Prices:")
print(df.isna().sum())
```

```python
print("Original Shape:", df.shape)

duplicate_count = df.duplicated().sum()

print("Duplicate Rows Found:", duplicate_count)

df = df.drop_duplicates()

print("New Shape:", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)

print("Updated ZIP Codes:")
print(df["zip"].head(10))

print("ZIP Code Data Type:")
print(df["zip"].dtype)
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    format="mixed"
)

print("Updated Column Types:")
print(df.dtypes)

display(df.head(10))
```

```python
negative_rows = df[df["qty"] < 0]

print("Negative Quantity Records:")
display(negative_rows)

print("Number of Negative Records:", len(negative_rows))
```

```python
# Keep negative quantities as a possible returns.
# No rows are removed and changed.

print("Negative quantities retained:", (df["qty"] < 0).sum())

print("Dataset Shape:", df.shape)
```

```python
# Negative Quantity
```

```python
df.to_csv("sales_clean.csv", index=False)

print("Cleaned sales data saved successfully!")

print("Final Shape:", df.shape)

display(df.head(10))
```

```python
print("FINAL DATA QUALITY CHECK")

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isna().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nColumn Data Types:")
print(df.dtypes)

print("\nZIP Code Lengths:")
print(df["zip"].str.len().value_counts())

print("\nNegative Quantities:")
print((df["qty"] < 0).sum())
```
