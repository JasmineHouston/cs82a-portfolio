```python
import pandas as pd

df = pd.read_csv("usarrests.csv")

df.head()
```

```python
df = df.rename(columns={"Unnamed: 0": "State"})

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
summary = df[["Murder", "Assault", "UrbanPop"]].describe()

summary
```

```python
extreme_assault = df[
    df["Assault"] > 500
]

extreme_assault
```

```python
low_urban = df[
    df["UrbanPop"] < 20
]

low_urban
```

```python
print("Missing values by column:")
print(df.isnull().sum())

print("\nRows with missing values:")
display(df[df.isnull().any(axis=1)])
```

```python
# Outlier  Analysis
# Flag 1: Iowa - UrbanPop = 570% (impossible)
# Urban population percentage cannot exceed 100%.
# Flag 2: South Carolina - Assault = 879 (extremely high)
# Flag 3: New York - UrbanPop = 6% (possible)
# For New York unusually low, could be a data entry error.
# Check original data entry error.
# Georgia Data Quality Issue - Missing Assault Value
# Data quality issue rather than an outlier.
```

```python
import pandas as pd

# Create a cleaning log
cleaning_log = pd.DataFrame({
    "State": [
        "Iowa",
        "South Carolina",
        "New York",
        "Georgia"
    ],
    "Column": [
        "UrbanPop",
        "Assault",
        "UrbanPop",
        "Assault"
    ],
    "Original Value": [
        570, 879, 6, None
    ],
    "Classification": [
        "Impossible",
        "Extreme",
        "Suspicious",
        "Missing"
    ],
    "Action": [
        "Verify before correction",
        "Investigate before changing",
        "Verify against source",
        "Keep missing until verified"
    ]
})

# Display the cleaning log
cleaning_log
```
