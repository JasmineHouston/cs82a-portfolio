```python
# Module 4 Lab 2: Bookstore Inventory Analysis

# Hands On Problem 3.1
# In this lab, I will use Python and pandas to analyze bookstore inventory data. My goal is to calculate the median book cover type, the average weight of all books, and the average weight of hardcover and paperback books separately. I will compare  the results and explain how this information could help make better inventory and shipping decisions.
```

```python
import pandas as pd

books = pd.read_csv("allbacks.csv")

print("First five rows:")
print(books.head())

print("\nColumn names:")
print(books.columns.tolist())

print("\nDataset information:")
print(books.info())
```

```python
# Median Cover Type

# The median is the middle value when data is arranged in order. Since book covers are categories instead of numbers, I will arrange paperback before hardback and determine the middle category. This will help me understand which cover type represents the center of the bookstore's inventory.
```

```python
cover_order = ["pb", "hb"]

covers = pd.Categorical(
    books["cover"].dropna(),
    categories=cover_order,
    ordered=True
)

sorted_covers = sorted(covers)

n = len(sorted_covers)

if n % 2 == 1:
    median_cover = sorted_covers[n // 2]
else:
    middle_two = sorted_covers[n // 2 - 1:n // 2 + 1]
    median_cover = (
        middle_two[0]
        if middle_two[0] == middle_two[1]
        else "Between pb and hb"
    )

print("Median cover type:", median_cover)
print("\nCover type counts:")
print(books["cover"].value_counts())
```

```python
# Mean Weight of All Books

# The mean is calculated by adding all the book weights together and dividing by the number of books with recorded weights. I will use pandas to calculate the average weight in grams. This information could help a bookstore estimate shipping costs and understand how heavy its inventory is overall.
```

```python
mean_weight = books["weight"].mean()

print("Mean weight of all books:",
      round(mean_weight, 2), "grams")
```

```python
# Mean Weight by Cover Type

# I will separate the books into hardcover and paperback groups and calculate the average weight for each group. Comparing these averages will help me determine which type of book is heavier and how that difference could affect inventory management and shipping.
```

```python
group_means = books.groupby("cover")["weight"].mean()

print("Average weight by cover type:")
print(group_means.round(2))

hardcover_mean = group_means["hb"]
paperback_mean = group_means["pb"]

difference = hardcover_mean - paperback_mean

print("\nHardcover average:",
      round(hardcover_mean, 2), "grams")

print("Paperback average:",
      round(paperback_mean, 2), "grams")

print("Difference:",
      round(abs(difference), 2), "grams")
```

```python
if hardcover_mean > paperback_mean:
    print("Hardcover books are heavier on average.")
elif paperback_mean > hardcover_mean:
    print("Paperback books are heavier on average.")
else:
    print("Both cover types have the same average weight.")

print("Average weight difference:",
      round(abs(difference), 2), "grams")
```

```python
# Interpretation

# After comparing the average weights, I found that Hardcover  books weigh more on average than Paperback books. The difference between the two groups is 168.31 grams. This information could help a bookstore estimate shipping costs, plan storage space, and manage inventory more efficiently. It also shows why comparing groups can provide more useful information than looking at one overall average.
```

```python
# Manual Verification

# To verify my results, I will calculate the average weight using the formula:

# Mean = Sum of all weights ÷ Number of books

# I will compare my manual calculation with the average calculated by pandas. This helps me confirm that the results are accurate and understand how the mean is calculated.
```

```python
total_weight = books["weight"].sum()
book_count = books["weight"].count()

manual_mean = total_weight / book_count

print("Total weight:", total_weight)
print("Number of books:", book_count)
print("Manual mean:", round(manual_mean, 2))
print("Pandas mean:", round(mean_weight, 2))

print("Results match:",
      abs(manual_mean - mean_weight) < 0.000001)
```

```python
summary = pd.DataFrame({
    "Statistic": [
        "Median Cover Type",
        "Mean Weight of All Books",
        "Mean Hardcover Weight",
        "Mean Paperback Weight",
        "Difference Between Group Means"
    ],
    "Result": [
        median_cover,
        f"{mean_weight:.2f} grams",
        f"{hardcover_mean:.2f} grams",
        f"{paperback_mean:.2f} grams",
        f"{abs(difference):.2f} grams"
    ]
})

display(summary)
```
