```python
# Module 4 Lab: Descriptive Statistics and Correlation

# Objective
# The purpose of this lab is to analyze cleaned sales data using descriptive statistics, histograms, and correlation analysis. I will examine product prices, quantities, and revenue to identify patterns and explain the results.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_clean.csv")

print("Dataset Shape:", df.shape)
print("\nFirst Five Rows:")
print(df.head())
```

```python
# Step 2: Center and Spread

# In this section, I will calculate the mean, median, and standard deviation for price and quantity. These statistics will help me understand typical values and how much the data varies.
```

```python
print("PRICE STATISTICS")
print("Mean:", df["price"].mean())
print("Median:", df["price"].median())
print("Standard Deviation:", df["price"].std())

print("\nQUANTITY STATISTICS")
print("Mean:", df["qty"].mean())
print("Median:", df["qty"].median())
print("Standard Deviation:", df["qty"].std())
```

```python
# Step 3: Which Summary Would I Report?

# I would report the median price because the dataset contains some higher priced items that can increase the average. The median gives a better representation of a typical product price when the data is right skewed.
```

```python
# Step 4: Price Distribution Histogram

# I will create a histogram to understand how product prices are distributed throughout the dataset.
```

```python
plt.figure(figsize=(8, 5))

df["price"].plot(
    kind="hist",
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Product Prices")
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Histogram Interpretation

# The price distribution appears right skewed, with most products having lower prices and fewer products having much higher prices. This supports using the median because expensive products can pull the mean upward.
```

```python
# Step 5: Calculate Revenue

# Revenue is calculated by multiplying the price of each product by the quantity sold. This creates a new column that can be used for further analysis.
```

```python
df["revenue"] = df["price"] * df["qty"]

print(df.head())
print("\nNew Dataset Shape:", df.shape)
```

```python
# Step 6: Correlation Analysis

# I will calculate the correlation between numeric variables to determine how strongly they are related. A positive correlation means the variables tend to increase together, while a negative correlation means they tend to move in opposite directions.
```

```python
correlation_matrix = df.corr(numeric_only=True)

print("Correlation Matrix:")
print(correlation_matrix.round(3))

print("\nPrice and Revenue Correlation:")
print(df["price"].corr(df["revenue"]))

print("\nQuantity and Revenue Correlation:")
print(df["qty"].corr(df["revenue"]))
```

```python
# Step 7: Scatter Plot Analysis

# I will compare price against revenue and quantity against revenue to see which variable has a stronger linear relationship with revenue.
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["price"],
    df["revenue"],
    alpha=0.6
)

plt.title("Price vs Revenue")
plt.xlabel("Price ($)")
plt.ylabel("Revenue ($)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["qty"],
    df["revenue"],
    alpha=0.6
)

plt.title("Quantity vs Revenue")
plt.xlabel("Quantity Sold")
plt.ylabel("Revenue ($)")
plt.grid(alpha=0.3)

plt.show()
```

```python
# Scatter Plot Comparison

# The scatter plots help me compare how price and quantity relate to revenue. I will use the correlation values and the patterns in both plots to determine which relationship is stronger. I will also check whether unusual values could influence the results.
```

```python
price_r = df["price"].corr(df["revenue"])
qty_r = df["qty"].corr(df["revenue"])

print(f"Price vs Revenue: r = {price_r:.3f}")
print(f"Quantity vs Revenue: r = {qty_r:.3f}")
```
