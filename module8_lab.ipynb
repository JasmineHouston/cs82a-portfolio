```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the penguins CSV.
df = pd.read_csv("penguins.csv")

# Show the number of missing values in each column.
print("Missing values by column:")
print(df.isna().sum())

# Compare the average flipper length for each species.
print("\nMean flipper length by species (mm):")
print(df.groupby("species")["flipper_length_mm"].mean())

# Plot flipper lengths by species. Seaborn omits rows missing either value.
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="species", y="flipper_length_mm")
plt.title("Flipper Length by Penguin Species")
plt.xlabel("Species")
plt.ylabel("Flipper length (mm)")
plt.tight_layout()
plt.show()
```
